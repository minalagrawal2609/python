```python
# Day - 5 Task: STUDENT DATA FILE MANAGER
import csv, json
from datetime import datetime

students= [
    {"name": "Ravi",  "marks": 88},
    {"name": "Priya", "marks": 45},
    {"name": "Arjun", "marks": 72},
    {"name": "Sneha", "marks": 91},
    {"name": "Karan", "marks": 55}
]

# get_grade function :-
def get_grade(marks):
    if marks >= 90: return "A+"
    elif marks >= 80: return "A"
    elif marks >= 70: return "B"
    elif marks >= 60: return "C"
    elif marks >= 50: return "D"
    else:             return "F"

# Step - 1 Write a writer_csv():-
def write_csv(students, filename):
  fieldnames = ["name", "marks", "grade", "result"]

  with open(filename, "w", newline="") as file:
      writer = csv.DictWriter(file, fieldnames=fieldnames)
      writer.writeheader()
      for s in students:
       writer.writerow({
          "name" : s["name"],
          "marks": s["marks"],
          "grade": get_grade(s["marks"]),
          "result": "Pass" if s["marks"] >=50 else "Fail"
       })

      print(f"[1] {filename} written  ({len(students)} records)")

# Step - 2 Read_csv():-
def read_csv(filename):
  print(f"[2] Reading from {filename}...")
  with open(filename, "r") as file:
    reader = csv.DictReader(file)
    for row in reader:
      print(f" {row['name'].ljust(8)}: {row['marks'].ljust(5)} Grade:{row['grade'].ljust(4)} {row['result']}")

# Step - 3 build_report() compute summary:-
def build_report(students):
    marks_list = [s["marks"] for s in students]
    passed     = [s for s in students if s["marks"] >=50]
    top        = max(students, key=lambda s: s["marks"])
    return{
        "class"       : "Python Batch A",
        "total"       : len(students),
        "passed"      : len(passed),
        "failed"      : len(students) - len(passed),
        "average"     : round(sum(marks_list)/ len(marks_list), 1),
        "top_scorer"  : f" {top['name']} ({top['marks']})",
        "generated at": datetime.now().strftime("%d/%m/%Y %H:%M:%S")
    }

# Step - 4 write_json():-
def write_json(report, filename):
  with open(filename, "w") as file:
       json.dump(report, file, indent=4)
       print(f"[3] {filename} written:-")

# Step - 5 Print_json_summary :- read and display
def print_json_summary(filename):
  with open(filename, "r") as file:
       data = json.load(file)
       print(f"[4] Summary from json:-")
       print(f"Class: {data['class']}")
       print(f"Total: {data['total']}")
       print(f"Passed: {data['passed']}")
       print(f"Failed: {data['failed']}")
       print(f"Average: {data['average']}")
       print(f"Top Scorer: {data['top_scorer']}")
       print(f"Generated at: {data['generated at']}")

# Step-6 run()-- all together
def run():
    print("=" * 30)
    print("  STUDENT DATA FILE MANAGER")
    print("=" * 30)
    csv_filename = "student_data.csv"
    json_filename = "report.json"

    write_csv(students, "students.csv")
    read_csv("students.csv")
    report = build_report(students)
    write_json(report, "report.json")
    print_json_summary("report.json")
    print("=" * 35)

run()
```
