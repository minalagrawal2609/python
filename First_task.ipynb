```python
# Day - 1 task: Student ID Card

# step - 1: Store the data
name = "kiran patel"
initials = "K . P"
age = 21
course = "computer science"
city = "ahmedabad"

# step - 2: Verify your types
print(type(name))
print(type(age))
print(type(course))
print(type(city))

#step - 3: Extract initials
print(name[0])     #'k'
print(name[6])     #'p'
print(name[0] + "." + name[6])

#step - 4: Slice the first name
print("name[0:5] =", name[0:5])

# step - 5: Reverse the full name
print("name[::-1]) =", name[::-1])

# step - 6: Alternate characters
print("alternate character[::2] =", course[::2])

# step - 7: Build ID card
print("===============================")
print("       STUDENT ID CARD")
print("===============================")
print("Name     :" + name)
print("Initials :" + initials)
print("Age      :" + str(age))
print("Course   :" + course)
print("city     :" + city)
print("==============================")











```
