```python
# Day - 4 Task: STUDENT REPORT GENERATOR

class_marks = {
    "Ravi" : 88,
    "Priya": 45,
    "Arjun": 72,
    "Sneha": 91,
    "Karan": 55
}

# #Step - 1 Write  a get_grade(marks) function and return:-
def get_grade(marks):
  if marks >= 90:
    return "A+"
  elif marks >= 80:
    return "A"
  elif marks >= 70:
    return "B"
  elif marks >= 60:
    return "C"
  elif marks >= 50:
    return "D"
  else:
    return "F"

# Step - 2 Write a print_student_rows(class_dict) function:-
def print_student_rows(class_dict):
    for name, marks in class_dict.items():
        grade = get_grade(marks)
        result = "PASS" if marks >= 50 else "FAIL"
        print(name, ":", marks, " Grade:", grade, "  ", result)

#Step - 6 print_report(class_dict):-
def print_report(class_dict):
    marks_list = list(class_dict.values())

#Step -3 list comprehension:-
    Passing= [m for m in marks_list if m >= 50]

#Step - 4 Map add 10% bonus to all marks, round to 1 daecimal:-
    bonus = list(map(lambda m: round(m * 1.1, 1), marks_list))

#Step _ 5 Filter - only marks above average
    average = sum(marks_list) / len(marks_list)
    above_avg = list(filter(lambda m: m > average, marks_list))

    print("==========================================")
    print("       STUDENT REPORT GENERATOR ")
    print("==========================================")
    print_student_rows(class_dict)
    print("========================================")
    print("Passing marks :", Passing)
    print("Marks + 10 bonus :", bonus)
    print("Above average :", above_avg)
    print("Average :", average)

print_report(class_marks)




```
