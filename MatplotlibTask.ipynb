DAY - 1 TASK OF MATPLOTLIB:-

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Create a NumPy array x of numbers from 0 to 10 using np.linspace(0, 10, 50).
fig, ax = plt.subplots()
x = np.linspace(0, 10, 50)

# Compute y = np.sin(x).
y = np.sin(x)

# Plot x vs y with a title and axis labels.
ax.plot(x, y, label="y = sin(x)")
ax.set_title("x vs y")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
plt.show()
```

```python
# Create a 1x3 subplot grid (plt.subplots(1, 3)).
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

# In the first chart, plot a line of y = x for x = np.arange(10).
x = np.arange(10)
y = x
axes[0].plot(x, y)
axes[0].set_title("line plot")
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")

# In the second chart, make a scatter plot of 20 random points (np.random.rand(20) for both x and y).
x = np.random.rand(20)
y = np.random.rand(20)
axes[1].scatter(x, y)
axes[1].set_title("Scatter plot")
axes[1].set_xlabel("x")
axes[1].set_ylabel("y")

# In the third chart, make a histogram of np.random.normal(0, 1, 200).
x = np.random.normal(0, 1, 200)
axes[2].hist(x)
axes[2].set_title("Histogram")
axes[2].set_xlabel("x")
axes[2].set_ylabel("y")

# Give each subplot its own title, and call plt.tight_layout() before plt.show().
plt.tight_layout()
plt.show()
```

DAY - 2 TASK OF MATPLOTLIB:-


```python
# Take the scatter plot of study_hours vs exam_score style data from Day 1 (or recreate similar random data).
np.random.seed(1)
study_hours = np.random.uniform(1, 10, 30)
exam_score = study_hours * 6 + np.random.normal(0, 8, 30)

fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(study_hours, exam_score, color="blue", alpha=0.6)
ax.set_title("Study hours vs Exam score")
ax.set_xlabel("Study hours")
ax.set_ylabel("Exam score")

Study_hours = np.max(study_hours)
Exam_score = np.max(exam_score)

ax.annotate("Highest score",
            xy=(Study_hours, Exam_score),
            xytext=(Study_hours + 1, Exam_score + 0.3),
            arrowprops=dict(facecolor="black", arrowstyle="->"))
fig.savefig("Scatter_plot.png", dpi=300)
plt.show()
```
