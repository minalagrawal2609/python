```python
from abc import ABC, abstractmethod

# step - 1 Abstract base class employees:-
class employees(ABC):
  def __init__(self, name, department, salary):
    self.name = name
    self.department = department
    self._salary = salary

#step - 5 Encapsulate (_salary):-
  @property
  def salary(self):
    return self._salary

  @salary.setter
  def salary(self, value):
    if value < 0:
      raise ValueError("Salary should be positive")
    self._salary = value

  @abstractmethod
  def calculate_salary(self):
    pass

  def print_details(self):
    print(f" Name: {self.name} Department: {self.department} Salary: {self.calculate_salary()}")

# STEP - 2 FullTimeEmployee(employees):-

class FullTimeEmployees(employees):
  def __init__(self, name, department, monthly_salary):
    super().__init__(name, department, monthly_salary)
    self.monthly_salary = monthly_salary

  def calculate_salary(self):
    return self.monthly_salary * 12

# step - 3 PartTimeEmployee(employees):-

class PartTimeEmployees(employees):
  def __init__(self, name, department, hourly_rate, hours_per_month):
    super().__init__(name, department, hourly_rate)
    self.hourly_rate = hourly_rate
    self.hours_per_month = hours_per_month

  def calculate_salary(self):
    return self.hourly_rate * self.hours_per_month * 12

# step - 4 ContractEmployee(employees):-
class ContractEmployees(employees):
  def __init__(self, name, department, contract_salary):
    super().__init__(name, department, contract_salary)
    self.contract_salary = contract_salary

  def calculate_salary(self):
    return self.contract_salary

# step - 6 Standalone print_report(employees) function:-
def print_report(employee_list):
  print("=========================================")
  print("    PYTHON ACADEMY — Employee Report")
  print("=========================================")

  totalpayroll = 0

  for e in employee_list:
    e.print_details()
    totalpayroll += e.calculate_salary()

  print("=========================================")
  print("Total employees:", (len(employee_list)))
  print("Total payroll:", totalpayroll)
  print("=========================================")

# step - 7 create atleast 3 employees
employee_list = [
      FullTimeEmployees("Rahul", "Engineering", 40000),
      PartTimeEmployees("Sneha", "Designer", 20, 500),
      ContractEmployees("Rajesh", "Finance", 55000)
  ]

print_report(employee_list)
```
