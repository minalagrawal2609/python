DAY - 1 Pandas task:-


```python
import pandas as pd
import numpy as np
```

```python
employees = pd.DataFrame({
    "EmpName": ["Ravi", "Meena", "Suresh", "Anita", "Vikram"],
    "Department": ["Sales", "IT", "IT", "HR", "Sales"],
    "Salary": [45000, 60000, 55000, 50000, 48000],
    "YearsExp": [2, 5, 3, 4, 1]
})
print(employees)
```

```python
# Display the first 3 rows of employees.
head = employees.head(3)
print(head)
```

```python
# Print the shape and column names of employees.
shape = employees.shape
print("Shape:", shape)
columns = employees.columns
print("Columns:", columns)
```

```python
# Select only the EmpName and Salary columns.
selected_columns = employees[["EmpName", "Salary"]]
print(selected_columns)
```

```python
# Filter to show only employees in the "IT" department.
IT_employees = employees[employees["Department"] == "IT"]
print(IT_employees)
```

```python
# Add a new column SeniorEmployee that is True if YearsExp >= 3, else False.
employees["SeniorEmployees"] = employees["YearsExp"] >= 3
print(employees)
```

```python
# Sort employees by Salary in descending order.
sorted_employees = employees.sort_values(by="Salary", ascending=False)
print(sorted_employees)
```

DAY - 2 PANDAS TASK:-

```python
orders = pd.DataFrame({
    "OrderID": [201, 202, 203, 204, 205, 206, 207],
    "Customer": ["Ravi", "Meena", "Ravi", "Anita", "Meena", "Vikram", "Anita"],
    "City": ["Delhi", "Mumbai", "Delhi", np.nan, "Mumbai", "Pune", "Chennai"],
    "Amount": [1200, np.nan, 800, 500, 1500, np.nan, 700]
})

print(orders)
```

```python
# Filter orders to show only orders from customer "Ravi" or "Meena" (use .isin()).
filter = orders[orders["Customer"].isin(["Ravi", "Meena"])]
print(filter)
```

```python
# Group orders by Customer and find the total Amount per customer (hint: NaN values are ignored automatically by .sum()).
group = orders.groupby("Customer")["Amount"].sum()
print(group)
```

```python
# Count how many missing values are in each column of orders
missing_values = orders.isna().sum()
print(missing_values)
```

```python
# Fill missing Amount values with the column's mean, and missing City values with "Unknown".
cleaned_orders = orders.copy()
cleaned_orders["Amount"] = orders["Amount"].fillna(orders["Amount"].mean())
cleaned_orders["City"] = orders["City"].fillna("Unknown")
print(cleaned_orders)
```

```python
# Save your cleaned orders DataFrame to a CSV file called orders_cleaned.csv.
cleaned_orders.to_csv("Orders.csv", index=False)
print("Saved cleaned_orders.csv")
```
