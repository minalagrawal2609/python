```python
# Day - 2 Task: CLASS GRADE BOOK REPORT

# Step - 1 Using Dictionary For Student Info
student = {
    "name"    : "Ravi Sharma",
    "roll_no" : 101,
    "subjects": ["Math", "Science", "English"],
    "marks"   : [88, 75, 92]
}

# Step - 2 Use a List for calculations
marks = student["marks"]
total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

# Step - 3 Use a set for passed subjects
subjects = student["subjects"]
passed = set()
passed.add(subjects[2]) if marks[2] >= 50 else None
passed.add(subjects[0]) if marks[0] >= 50 else None
passed.add(subjects[1]) if marks[1] >= 50 else None

# Step - 4 Use Tuple for school info
school_info = [" Greenfield school", "Ahmedabad", 2005]

# Step - 5 Print the grade book
print("===============================")
print("   CLASS GRADE BOOK REPORT")
print("===============================")
print("Student  :" + student["name"])
print("Roll No  : "+ str(student["roll_no"]))
print("Subjects :" + str(student["subjects"]))
print("Marks    :" + str(marks))
print("Total    :" + str(total))
print("Average  :" + str(average))
print("Highest  :" + str(highest))
print("Lowest   :" + str(lowest))
print("Subject passed :" + str(passed))
print("School info :" + str(school_info))
print("===============================")
```
