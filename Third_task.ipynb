```python
class_marks = {
    "Ravi" : 88,
    "Priya": 45,
    "Arjun": 72,
    "Sneha": 91,
    "Karan": 55
}

# Step -1 to 4 One loop to compute everthing

total = 0
pass_count = 0
fail_count = 0
top_name = ""
top_score = 0

print("===============================")
print("   CLASSROOM GRADE ANALYZER")
print("===============================")

# Step - 1 loop and assign a grade
for name, marks in class_marks.items():

  if marks >= 90:
    grade = "A+"
  elif marks >= 80:
    grade = "A"
  elif marks >= 70:
    grade = "B"
  elif marks >= 60:
    grade = "C"
  elif marks >= 50:
    grade = "D"
  else:
    grade = "F"

# Step - 2 count Pass/Fail
  if marks >= 50:
    Result = "Pass"
    pass_count = pass_count + 1
  else:
    Result = "Fail"
    fail_count = fail_count + 1

# Step - 3 find the total (for all students)
  total = total + marks

# Step - 4 top scorer (for all students)
  if marks > top_score:
    top_name = name
    top_score = marks

# Print each student's result
  print(name, ":", marks, " —", " Grade :", grade, ":-", Result)

# Step - 5 Print summary statistics AFTER the loop
print("===============================")
print("Total student :", len(class_marks))
print("Passed        :", pass_count)
print("Failed        :", fail_count)
print("Class average :", total / len(class_marks))
print("Top scorer    :", top_score, "with",top_name)
print("===============================")
```
